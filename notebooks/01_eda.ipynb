```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../data/sample/acme_corp.csv")
df["period"] = pd.to_datetime(df["period"] + "-01")
df.set_index("period", inplace=True)
df.head()
```

```python
df.describe()
```

```python
df[["revenue", "expenses", "net_profit", "operating_cash_flow"]].plot(
    figsize=(12, 5),
    title="Acme Corp — Financial Trends",
)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Feature Correlation Matrix")
plt.show()
```
